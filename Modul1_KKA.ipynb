```python
class Kucing:
 # Constructor
    def __init__(self, nama, warna):
      self.nama = nama # Attribute
      self.warna = warna # Attribute
 # Method
    def suara(self):
      return f"{self.nama} berkata: Meow!"

# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih")
kucing2 = Kucing("Garfield", "Oranye")
# Mengakses Attribute dan Method
print(kucing1.nama) # Output: Luna
print(kucing1.warna) # Outpur : Putih
print(kucing1.suara()) # Output: Luna berkata: Meow!
print(kucing2.nama) # Output: Garfield
print(kucing2.warna) # Output : Oranye
print(kucing2.suara()) # Output : Garfield berkata: Meow
```

```python
class AkunBank:
  def __init__(self, saldo):
    self.__saldo = saldo # Private attribute

  def cek_saldo(self):
    return self.__saldo

  def setor(self, jumlah):
    if jumlah > 0:
      self.__saldo += jumlah

  def tarik(self, jumlah):
    if jumlah > 0 and jumlah <= self.__saldo:
      self.__saldo -= jumlah
    else:
      print("Jumlah yang dimasukkan tidak valid atau saldo tidak mencukupi.")

rekening = AkunBank(1000)
rekening.setor(500)
rekening.setor(1000)
rekening.tarik(1500)
print(rekening.cek_saldo()) # Output: 1500
# print(rekening.__saldo) # Ini akan Error! Tidak bis
```

```python
# ==========================================
# LATIHAN 1: MEMBUAT CLASS HERO
# ==========================================

class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name          # Nama Hero
        self.hp = hp              # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Percobaan Tugas Analisis 1: Mengubah HP secara langsung
hero1.hp = 500

# Memanggil Method dan Print
hero1.info()
hero2.info()
print(f"HP Hero 1 setelah diubah: {hero1.hp}")
```

**TUGAS ANALISIS 1**

Jika hero1.hp diubah menjadi 500, maka nilai HP pada object hero1 berubah menjadi 500. Hal ini terjadi karena hp merupakan attribute yang bisa diakses dan diubah secara langsung dari luar class.

```python
# ==========================================
# LATIHAN 2: INTERAKSI ANTAR OBJEK
# ==========================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

**TUGAS ANALISIS 2**

Parameter lawan penting karena parameter tersebut menerima object hero secara langsung. Dengan begitu, method serang() dapat mengakses data dan method milik hero yang diserang, seperti lawan.name dan lawan.diserang(). Jadi object dapat berinteraksi secara langsung dengan object lainnya.

```python
# ==========================================
# LATIHAN 3: PEWARISAN (INHERITANCE)
# ==========================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} (Mage) | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)         # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

**TUGAS ANALISIS 3**

Apa yang terjadi jika super().__init__() dihapus?

Jika super().__init__(name, hp, attack_power) dihapus, constructor dari class Hero tidak akan dijalankan. Akibatnya attribute seperti name, hp, dan attack_power tidak dibuat pada object Mage. Ketika eudora.info() dijalankan, program akan menghasilkan error seperti AttributeError karena object Mage tidak memiliki attribute name.

Apa peran super()?

super() digunakan untuk memanggil method dari class induk. Pada program ini, super().__init__() digunakan agar attribute name, hp, dan attack_power dari class Hero tetap dibuat ketika object Mage dibuat. Jadi class Mage tidak perlu menulis ulang kode constructor dari Hero.

```python
# ==========================================
# LATIHAN 4: ENKAPSULASI
# ==========================================

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: Private Attribute (__hp)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0 # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)

hero1.__hp = 9999     # GAGAL (Menciptakan variabel baru di luar class)
hero1.set_hp(-50)     # Uji coba nilai negatif
print(f"HP Hero 1: {hero1.get_hp()}") # Output: 0

# Percobaan Akses Paksa (Name Mangling)
print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

Percobaan Hacking (_Hero__hp):  Hasil: Nilai HP muncul.  
Penjelasan: Python menggunakan mekanisme Name Mangling, di mana atribut diawali __ diubah namanya menjadi _NamaClass__namaAtribut secara internal. Python tidak sepenuhnya melarang akses, tetapi mengandalkan kedisiplinan programmer. Kita tetap tidak boleh melakukannya karena merusak prinsip Enkapsulasi dan dapat menyebabkan bug jika struktur internal class berubah.  

Uji Validasi (Atribut tanpa Logika if/elif):  
Hasil set_hp(-100): HP hero akan menjadi -100.  
Pentingnya Setter: Method Setter menjamin Integritas Data agar nilai atribut tetap masuk akal sesuai logika bisnis sistem/game (misalnya: HP tidak boleh minus atau bernilai terlampau besar akibat cheat).  

```python
# ==========================================
# LATIHAN 5: ABSTRACTION & INTERFACE
# ==========================================

from abc import ABC, abstractmethod

# 1. Interface / Abstract Class (KONTRAK)
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()

# Menguji coba pembuatan objek dari Abstract Class langsung
try:
    unit = GameUnit()
except TypeError as e:
    print(f"\n[ERROR TERDETEKSI]: {e}")
```

**TUGAS ANALISIS 5**

Melanggar Kontrak (Hapus method serang di Hero):  Pesan Error: TypeError: Can't instantiate abstract class Hero with abstract method serang.  

Penjelasan: Class Hero gagal dijadikan objek karena belum mengimplementasikan semua method abstrak yang telah diwajibkan oleh GameUnit. Jika kita lupa membuat method tersebut, program akan menolak pembuatan objek saat runtime.

Mencetak Cetakan (GameUnit()):  Alasan Dilarang: Class abstrak hanyalah sebuah gambaran/kontrak mentah dan belum memiliki fungsi/implementasi nyata.  

Kegunaan: Sebagai cetakan standar (blueprint) yang memaksakan aturan kepada semua class turunan agar memiliki metode dan struktur interaksi yang konsisten.  

```python
# ==========================================
# LATIHAN 6: POLYMORPHISM
# ==========================================

class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# Tugas Analisis 6.1: Menambahkan class Healer
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")

# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Rafaela") # Ditambahkan ke dalam list
]

print("\n--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

**TUGAS ANALISIS 6**

**Uji Skalabilitas:**  Hasil: Program berjalan lancar tanpa merusak/mengubah kode pengulangan (looping).  

**Kesimpulan:** Polimorfisme mempermudah pengembangan sistem secara modular (Scalable). Programmer dapat menambahkan fitur/karakter baru tanpa perlu memodifikasi kode utama atau logika eksekusi yang sudah ada.  

**Konsistensi Penamaan:**  Hasil jika diubah ke tembak_panah: Program akan memanggil method serang() dari class induk Hero untuk Archer (mengeluarkan output "Hero menyerang dengan tangan kosong.").  

**Alasan:** Polimorfisme bekerja berdasarkan kesamaan antarmuka (nama method yang sama). Jika nama method berbeda, program menganggap method tersebut tidak me-override method milik class induk.  

```python
from abc import ABC, abstractmethod


# =========================================================
# ABSTRACT CLASS
# =========================================================

class KamarHotel(ABC):
    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = 0
        self.__harga_dasar = harga_dasar

        # Menggunakan setter untuk validasi stok awal
        self.tambah_stok(stok)

    # Getter stok
    def get_stok(self):
        return self.__stok

    # Getter harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Method untuk menambah/mengubah stok
    def tambah_stok(self, jumlah):
        if self.__stok + jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah

    # Abstract method
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# =========================================================
# CHILD CLASS: KAMAR DELUXE
# =========================================================

class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    # Polymorphism / Override
    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 10 / 100

        print(
            f"[DELUXE] {self.nama_kamar} | "
            f"Fasilitas: {self.fasilitas}"
        )
        print(
            f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | "
            f"Pajak(10%): Rp {pajak:,.0f}"
        )

    # Polymorphism / Override
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 10 / 100

        return (harga_dasar + pajak) * jumlah_malam


# =========================================================
# CHILD CLASS: KAMAR STANDARD
# =========================================================

class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    # Polymorphism / Override
    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 5 / 100

        print(
            f"[STANDARD] {self.nama_kamar} | "
            f"Kapasitas: {self.kapasitas}"
        )
        print(
            f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | "
            f"Pajak(5%): Rp {pajak:,.0f}"
        )

    # Polymorphism / Override
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 5 / 100

        return (harga_dasar + pajak) * jumlah_malam


# =========================================================
# FUNGSI PROSES TRANSAKSI
# =========================================================

def proses_transaksi(daftar_pesanan):
    total_tagihan = 0

    print("\n--- STRUK PEMESANAN ---")

    for nomor, pesanan in enumerate(daftar_pesanan, start=1):
        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["malam"]

        kamar.tampilkan_detail()

        subtotal = kamar.hitung_harga_total(jumlah_malam)

        print(
            f"Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {subtotal:,.0f}"
        )
        print()

        total_tagihan += subtotal

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("----------------------------------------")


# =========================================================
# MAIN PROGRAM
# =========================================================

print("--- SETUP DATA KAMAR ---")

# Admin membuat kamar
deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1_500_000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500_000,
    "2 Orang"
)


# Admin menambahkan stok
deluxe.tambah_stok(10)
print("Berhasil menambahkan stok Kamar Deluxe: 10 unit.")


# Mencoba memasukkan stok negatif
standard.tambah_stok(-5)


# Menambahkan stok standard
standard.tambah_stok(20)
print("Berhasil menambahkan stok Kamar Standard: 20 unit.")


# =========================================================
# PEMESANAN
# =========================================================

daftar_pesanan = [
    {
        "kamar": deluxe,
        "malam": 2
    },
    {
        "kamar": standard,
        "malam": 1
    }
]


# Memproses transaksi
proses_transaksi(daftar_pesanan)
```
